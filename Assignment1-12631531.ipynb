```python
import numpy as np

#2.1
#创建矩阵
M1 = np.random.randint(0, 51, size=(5, 10))
M2 = np.random.randint(0, 51, size=(10, 5))

#输出创建的两个矩阵
print(M1)
print(M2)

#2.2 编写矩阵乘法函数
def Matrix_multip(matrix1, matrix2):
    #获取矩阵大小
    Row_1 = len(matrix1) #矩阵1的行数
    Col_1 = len(matrix1[0]) #矩阵1的列数
    Row_2 = len(matrix2) #矩阵2的行数
    Col_2 = len(matrix2[0])
    
    #判断能否相乘
    if Col_1 == Row_2:
        #创建一个全零矩阵
        matrix0 = np.zeros((Row_1, Col_2))
        #矩阵M1乘以矩阵M2
        for i in range(len(matrix1)):
            for j in range(len(matrix2[0])):
                for k in range(len(matrix2)):
                    matrix0[i][j] += matrix1[i][k] * matrix2[k][j]
        return matrix0
    else:
        return None 


#调用函数计算M1 * M2
M3 = Matrix_multip(M1, M2)

# 只有当 M3 不是 None（即成功相乘）时，才打印结果
if M3 is not None:
    print(M3)
else:
    print("矩阵无法相乘！")

```

```python
#定义帕斯卡三角函数
def Pascal_Triangle(k):
    #定义初始行
    Row = [1]
    
    #算出k行的列数
    for n in range(1, k + 1):
        New_Row = [1] * (n + 1)
        for i in range(1, n):
            New_Row[i] = Row[i - 1] + Row[i]
        Row = New_Row
    return Row

#调用函数显示帕斯卡三角第100 & 200行
print(Pascal_Triangle(100))
print(Pascal_Triangle(200))
```

```python
def Least_moves(x):
    #初始金额
    RMB = x
    
    #金额数值转化为二进制形式
    Binary = bin(RMB)[2:]
    #print(Binary)
    
    #计算二进制数值的位数
    Digit_Binary = len(Binary)
    #print(len(Binary))
    
    #计算二进制数值中1的个数
    Count = Binary.count('1')
    #print(Count)
            
    #得到步数        
    Step = Digit_Binary + Count - 2
    return Step

print(Least_moves(5))
```

```python
#5.1 编写函数Find_expression

#定义递归函数
def Find_expression(k):
    #初始数字
    Digit = "123456789"
    
    #终止条件
    def dfs(idex, output):
        if idex == 8:
            if eval(output) == k:
               print(f"{output} = {k}")
            return
    
        #递归条件
        New_Digit = Digit[idex + 1]
        dfs(idex + 1, output + "+" + New_Digit)
        dfs(idex + 1, output + "-" + New_Digit)
        dfs(idex + 1, output + New_Digit)
        
    #初始条件
    dfs(0, Digit[0])

#选取k=50进行验证
Find_expression(50)


#5.2 编写函数Count_solutions，并找到最值解及数字      
#定义递归函数
def Count_solutions(k):
    #初始数字
    Digit = "123456789"
    #计数
    Visited = 0
    
    #终止条件
    def dfs(idex, output):
        nonlocal Visited
        if idex == 8:
            if eval(output) == k:
               Visited += 1
            return
        
        #递归条件
        New_Digit = Digit[idex + 1]
        dfs(idex + 1, output + "+" + New_Digit)
        dfs(idex + 1, output + "-" + New_Digit)
        dfs(idex + 1, output + New_Digit)
    
    #初始条件
    dfs(0, Digit[0])
    #计数
    return Visited

# 统计 1 到 100解个数
Total_solutions = [Count_solutions(k) for k in range(1, 101)]

# 求最值解
Max_count = max(Total_solutions)
Min_count = min(Total_solutions)
print(Max_count)
print(Min_count)

#求最值解对应数字
for i, c in list(enumerate(Total_solutions)):
    if c == max(Total_solutions):
        print(f"最大值对应数字：{i + 1}")
    elif c == min(Total_solutions):
        print(f"最小值对应数字：{i + 1}")
#助教添加
import matplotlib.pyplot as plt

numbers = range(1, 101)

plt.plot(numbers, Total_solutions)
plt.xlabel("Number")
plt.ylabel("Total solutions")
plt.title("Total Solutions for Numbers from 1 to 100")
plt.show()
```
